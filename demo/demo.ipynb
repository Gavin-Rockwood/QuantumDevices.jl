## Test Model

```julia
q1 = make_qubit("q1", 1.0)
q2 = make_qubit("q2", 1.0)
interaction = op(:q1_x) * op(:q2_x) + op(:q1_y) * op(:q2_y)

model = make_model([q1, q2], param(:g)*interaction, (g = 0.0,))
model
```

```julia
model.parameters
```

### Some Gates

#### Single Qubit Gates

##### X1

```julia
duration = 10
gate_params = (drive = gaussian_pulse(0.1, 4), )
drive_term = param(:drive)*op(:q1_x)

X1_raw = DeviceGate(gate_params, drive_term, duration)
X1_target = tensor(sigmax(), qeye(2))
```

```julia
parameters(X1_raw)
```

```julia
# Parameter discovery is now provided by QuantumDevices.
parameters(X1_raw)
```

```julia
setup = calibration_problem(model, X1_raw, ["drive/amplitude"], X1_target)
```

```julia
res = calibrate(setup, NelderMead())
```

```julia
parameters(res[2])
```
